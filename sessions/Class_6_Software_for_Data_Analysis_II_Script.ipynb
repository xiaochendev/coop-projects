# 7. Histograms and Distributions

A histogram is different from a bar chart.

## Bar chart
- compares categories
- each bar represents a category

## Histogram
- shows the distribution of a numerical variable
- each bar represents a range of numerical values, called a **bin**

In this section, we will create a new numerical variable: **Response Hours**.

````python
# Convert date columns using format="mixed" to suppress the warning and speed up parsing
df["Created Date"] = pd.to_datetime(
    df["Created Date"],
    format="mixed",
    errors="coerce"
)

df["Closed Date"] = pd.to_datetime(
    df["Closed Date"],
    format="mixed",
    errors="coerce"
)
````

`pd.to_datetime()` converts date-like text into Python datetime values.

`errors="coerce"` changes values Python cannot interpret as dates into missing values instead of stopping the entire program.

````python
# Create a response-time variable in hours

df["Response Hours"] = (
    df["Closed Date"] - df["Created Date"]
).dt.total_seconds() / 3600

df[["Created Date", "Closed Date", "Response Hours"]].head()
````


---

### 💡 Understanding Datetime Conversions & Missing Data

When we load raw datasets like NYC 311 complaints, date fields are initially imported as general text strings (e.g., `"08/31/2023 11:59:00 PM"`).

To analyze time trends or calculate how long a ticket took to close, we use `pd.to_datetime()` to convert these strings into official **pandas Datetime objects**.

---

#### Key Parameters in `pd.to_datetime()`

* **`format="mixed"`**
* **What it does:** Tells pandas that the date strings might follow slightly different visual layouts across different rows.
* **Why we use it:** It suppresses unnecessary warnings and speeds up conversion by preventing pandas from guessing row-by-row.


* **`errors="coerce"`**
* **What it does:** Instructs pandas on what to do if it encounters blank cells, corrupted text, or unparseable dates.
* **Why we use it:** Instead of crashing your code with an error, it safely converts invalid values into a special missing-data placeholder called **`NaT`**.



---

#### `NaN` vs. `NaT`: What is the difference?

When working with data analysis in Python, you will encounter two main missing-value markers:

| Marker | Full Name | Applies To | What it means in our 311 dataset |
| --- | --- | --- | --- |
| **`NaN`** | **N**ot **a** **N**umber | Numeric values, text, & general objects | A ticket is missing a ZIP code, incident address, or borough name. |
| **`NaT`** | **N**ot **a** **T**ime | `datetime` and `timedelta` objects | A ticket has no `"Closed Date"` because it is **still open** or under investigation. |

> **Pro-Tip:** Functions like `df.isna()` or `df.isnull()` catch both `NaN` and `NaT` automatically!

## Box Plot of Response Hours — With Outliers

For this visualization, we are using **Matplotlib**, a Python library used to create charts and graphs. We call Matplotlib with `import matplotlib.pyplot as plt`. The abbreviation `plt` makes it easier to refer to Matplotlib throughout our code.

A **box plot** helps us understand the distribution of a numeric variable. Here, we are examining how long it took to respond to 311 service requests, measured in **hours**.

The box plot shows:

- **Median** — the line inside the box represents the middle value.
- **Box** — represents the middle 50% of the response times.
- **Whiskers** — show the typical range of the data.
- **Outliers** — individual points beyond the whiskers that represent unusually high or low response times.

In the code below:

- `import matplotlib.pyplot as plt` loads Matplotlib's plotting tools.
- `df["Response Hours"]` selects the Response Hours column from our pandas DataFrame.
- `.dropna()` removes missing values before creating the graph.
- `plt.boxplot()` tells Matplotlib to create a box plot.
- `showfliers=True` tells Matplotlib to display potential outliers.
- `plt.show()` displays the completed visualization.

### Question to Consider

What do the outliers tell us about 311 response times? Why might some requests take much longer than others?

````python
import matplotlib.pyplot as plt

# Create a box plot with outliers
plt.boxplot(
    df["Response Hours"].dropna(),
    showfliers=True,
    tick_labels=["Response Time"]
)

# Add title and axis label
plt.title("Box Plot of 311 Response Hours")
plt.ylabel("Hours")

# Display the plot
plt.show()
````

## Box Plot of Response Hours — Without Displaying Outliers

We are again using **Matplotlib** to create our visualization. Remember that **pandas helps us organize and analyze the data**, while **Matplotlib helps us visualize the data**.

Extreme values can sometimes make the main part of a box plot difficult to see. We can create the same box plot while **hiding the outlier points** to get a clearer view of typical response times.

In the code below:

- `df["Response Hours"]` selects the Response Hours column.
- `.dropna()` removes missing values before creating the graph.
- `plt.boxplot()` creates the box plot using Matplotlib.
- `showfliers=False` hides potential outliers from the visualization.
- `plt.show()` displays the completed visualization.
- The underlying data **does not change** — the outliers are only hidden from the graph.

Comparing the two graphs helps us see how **outliers can affect the scale and appearance of a visualization**.

### Question to Consider

Compare the two box plots. Which graph makes it easier to see the typical range of 311 response times? What information do we lose when we hide the outliers?

````python
import matplotlib.pyplot as plt

# Create a box plot without displaying outliers
plt.boxplot(
    df["Response Hours"].dropna(),
    showfliers=False,
    tick_labels=["Response Time"]
)

# Add title and axis label
plt.title("Box Plot of 311 Response Hours")
plt.ylabel("Hours")

# Display the plot
plt.show()
````

````python
# Get the response hours data
response_hours = df["Response Hours"].dropna()

# Calculate summary statistics
minimum = response_hours.min()
q1 = response_hours.quantile(0.25)
median = response_hours.median()
mean = response_hours.mean()
q3 = response_hours.quantile(0.75)
maximum = response_hours.max()

# Create the box plot
fig, ax = plt.subplots(figsize=(8, 6))

ax.boxplot(
    response_hours,
    showfliers=False,
    tick_labels=["Response Time"]
)

# Add title and axis label
ax.set_title("Box Plot of 311 Response Hours")
ax.set_ylabel("Hours")

# Create statistics label
stats_text = (
    f"Minimum: {minimum:.1f} hrs\n"
    f"Q1: {q1:.1f} hrs\n"
    f"Median: {median:.1f} hrs\n"
    f"Mean: {mean:.1f} hrs\n"
    f"Q3: {q3:.1f} hrs\n"
    f"Maximum: {maximum:.1f} hrs"
)

# Add statistics beside the chart
ax.text(
    1.05, 0.5,
    stats_text,
    transform=ax.transAxes,
    verticalalignment="center"
)

# Add footnote
fig.text(
    0.5, 0.02,
    "Note: Outliers are hidden. The whiskers do not necessarily represent the actual minimum and maximum values.",
    ha="center",
    fontsize=9
)

# Adjust spacing
plt.subplots_adjust(right=0.75, bottom=0.15)

plt.show()
````

## Line Chart: Daily 311 Service Requests

A **line chart** is useful for showing how data changes over time. Here, we are looking at the number of 311 service requests received each day during August 2026.

### What Does the Code Do?

- `daily_counts.plot(kind="line")` → creates the line chart.
- `plt.title()` → adds a title.
- `plt.xlabel()` and `plt.ylabel()` → label the axes.
- `DayLocator(interval=5)` → shows a date label every five days.
- `DateFormatter("%b %d")` → displays dates as **Aug 01, Aug 06, Aug 11**, etc.
- `plt.xticks(rotation=45)` → rotates the dates so they are easier to read.
- `plt.grid()` → adds grid lines to help us read the chart.
- `plt.show()` → displays the chart.

### What Should We Look For?

Use the line chart to look for **trends, increases, decreases, and unusually busy days** in 311 service requests.

````python
import matplotlib.dates as mdates

# Create the line chart
daily_counts.plot(kind="line", figsize=(10, 5))

# Add title and axis labels
plt.title("Daily 311 Service Requests — August 2026")
plt.xlabel("Date")
plt.ylabel("Number of Service Requests")

# Clean up the x-axis
ax = plt.gca()
ax.xaxis.set_major_locator(mdates.DayLocator(interval=5))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %d"))

# Rotate date labels
plt.xticks(rotation=45)

# Add horizontal grid lines
plt.grid(axis="y", alpha=0.3)

# Adjust spacing
plt.tight_layout()

plt.show()
````

````python
# Find the lowest-volume day

lowest_day = daily_counts.idxmin()
lowest_count = daily_counts.min()

lowest_day, lowest_count
````

## Interpretation Caution

If one day has the highest count, we can say:

> It was the highest-volume day in this dataset.

We should not automatically call it an **outlier** or say it was **statistically unusual** without additional analysis.

"Highest" and "statistically unusual" are different claims.

# 9. Design Basics and Chart Junk

Correct code does not automatically create a good chart.

A strong chart should make the data easier to understand.

## Good Design Principles

- use a clear title
- use readable labels
- use an appropriate scale
- keep color simple and purposeful
- reduce clutter
- emphasize important information
- remove decoration that does not help explain the data

## Chart Junk

**Chart junk** is visual material that distracts from the data without helping communicate the message.

Examples:
- unnecessary 3-D effects
- gradients
- excessive colors
- decorative icons
- unnecessary borders
- too many gridlines
- tiny labels
- complicated legends

# The Five-Second Test

A useful visualization check is the **five-second test**:

1. Look at a chart for about five seconds.
2. Look away.
3. Ask yourself: **What was that chart about?**

If the basic message is difficult to remember, the chart may need revision.

A useful design principle is:

> **Less clutter → more attention on the data**

# 12. Writing an Insight Statement

A chart is not always the finished product. Analysts also need to explain the main result.

A simple framework:

## 1. What happened?
State the main finding.

## 2. Evidence
Give the number, comparison, or pattern.

## 3. Why it matters
Explain why the audience should care, without claiming more than the data supports.

### Weak example

> Brooklyn is high.

### Better example

> Brooklyn recorded the highest number of August 311 service requests among the five boroughs.

### Stronger example

> Brooklyn recorded the highest August 311 volume among the five boroughs, indicating that a citywide review of service demand should examine the borough's complaint mix closely.

Be careful not to claim that a chart proves **why** a pattern occurred unless the analysis supports that conclusion.
