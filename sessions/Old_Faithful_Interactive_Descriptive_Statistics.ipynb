## 5. Descriptive statistics with `describe()`

The `describe()` function calculates several common descriptive statistics automatically.

These include:

- **count**
- **mean**
- **standard deviation**
- **minimum**
- **25th percentile**
- **median (50th percentile)**
- **75th percentile**
- **maximum**


```python
df.describe().round(2)
```

## 6. Calculate individual descriptive statistics

Sometimes we want to calculate statistics separately so that we can discuss what each one means.


```python
print("ERUPTION DURATION")

print("Count:", df["eruptions"].count())
print("Mean:", round(df["eruptions"].mean(), 2))
print("Median:", round(df["eruptions"].median(), 2))
print("Mode:", round(df["eruptions"].mode().iloc[0], 2))
print("Minimum:", round(df["eruptions"].min(), 2))
print("Maximum:", round(df["eruptions"].max(), 2))
print("Range:", round(df["eruptions"].max() - df["eruptions"].min(), 2))
print("Variance:", round(df["eruptions"].var(), 2))
print("Standard Deviation:", round(df["eruptions"].std(), 2))
```

```python
print("WAITING TIME")

print("Count:", df["waiting"].count())
print("Mean:", round(df["waiting"].mean(), 2))
print("Median:", round(df["waiting"].median(), 2))
print("Mode:", round(df["waiting"].mode().iloc[0], 2))
print("Minimum:", round(df["waiting"].min(), 2))
print("Maximum:", round(df["waiting"].max(), 2))
print("Range:", round(df["waiting"].max() - df["waiting"].min(), 2))
print("Variance:", round(df["waiting"].var(), 2))
print("Standard Deviation:", round(df["waiting"].std(), 2))
```

### Class discussion

- How are the **mean** and **median** different?
- Which variable has the larger range?
- Which variable has the larger standard deviation?
- What does standard deviation tell us?
- Why does variance have squared units while standard deviation uses the original units?


## 7. Create a summary table

We can combine several statistics into one clean comparison table.


```python
summary_table = pd.DataFrame({
    "Mean": df.mean(),
    "Median": df.median(),
    "Minimum": df.min(),
    "Maximum": df.max(),
    "Range": df.max() - df.min(),
    "Variance": df.var(),
    "Standard Deviation": df.std()
})

summary_table.round(2)
```

## 8. Interactive histogram

A **histogram** shows the distribution of a numeric variable.

It groups values into intervals called **bins** and counts how many observations fall into each interval.


```python
print("Choose a variable:")
print("1 = eruptions")
print("2 = waiting")

choice = input("Enter 1 or 2: ")

if choice == "1":
    variable = "eruptions"
elif choice == "2":
    variable = "waiting"
else:
    print("Invalid choice. Using eruptions.")
    variable = "eruptions"

bins = input("How many bins would you like? Press Enter for 10: ")

if bins == "":
    bins = 10
else:
    bins = int(bins)

plt.figure(figsize=(9, 5))
plt.hist(df[variable], bins=bins, edgecolor="black")

plt.title(f"Distribution of {variable.title()}")
plt.xlabel(variable.title())
plt.ylabel("Frequency")

plt.show()
```

### Histogram discussion

Look at the shape of the distribution.

Ask:

1. Where are most observations concentrated?
2. Is the distribution symmetric?
3. Is it skewed?
4. Does it appear to have more than one cluster?
5. Are there any gaps?
6. Are there possible outliers?

The Old Faithful data often shows distinct clusters, which can lead to a useful discussion of **bimodal distributions**.

## 11. Create an observation number

This dataset does **not** contain a date or time variable.

For a line chart, we can create an observation number:

1, 2, 3, 4, ... through the final observation.

### Important

This means the line chart shows the **sequence of observations**, not change over calendar time.


```python
df["observation"] = range(1, len(df) + 1)

df.head()
```

## 12. Interactive line chart

Choose which numeric variable you would like to plot across observation order.


```python
print("Choose a variable for the line chart:")
print("1 = eruptions")
print("2 = waiting")

choice = input("Enter 1 or 2: ")

if choice == "1":
    variable = "eruptions"
elif choice == "2":
    variable = "waiting"
else:
    print("Invalid choice. Using eruptions.")
    variable = "eruptions"

plt.figure(figsize=(11, 5))

plt.plot(
    df["observation"],
    df[variable],
    marker="o",
    markersize=3
)

plt.title(f"{variable.title()} by Observation Number")
plt.xlabel("Observation Number")
plt.ylabel(variable.title())

plt.show()
```

### Line chart discussion

Ask students:

1. Do you see alternating periods of higher and lower values?
2. Are there sudden jumps?
3. Are values clustered?
4. Does the pattern appear random or structured?
5. Why should we **not** describe this as a trend over time?

Remember: the x-axis is observation order, not dates.

```python
plt.figure(figsize=(11, 5))

plt.plot(
    df["observation"],
    df["waiting"]
)

plt.title("Waiting Time by Observation Number")
plt.xlabel("Observation Number")
plt.ylabel("Waiting Time (minutes)")

plt.show()
```

## 15. Final comparison table

Let's compare the two original numeric variables one more time.


```python
comparison = pd.DataFrame({
    "Mean": df[["eruptions", "waiting"]].mean(),
    "Median": df[["eruptions", "waiting"]].median(),
    "Minimum": df[["eruptions", "waiting"]].min(),
    "Maximum": df[["eruptions", "waiting"]].max(),
    "Range": (
        df[["eruptions", "waiting"]].max()
        - df[["eruptions", "waiting"]].min()
    ),
    "Standard Deviation": df[["eruptions", "waiting"]].std()
})

comparison.round(2)
```
