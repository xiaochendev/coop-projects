## 6. Calculate individual descriptive statistics

Sometimes we want to calculate statistics separately so that we can discuss what each one means.


```python
print("Sum of Dec")

print("Count:", df["Sum of Dec"].count())
print("Mean:", round(df["Sum of Dec"].mean(), 2))
print("Median:", round(df["Sum of Dec"].median(), 2))
print("Mode:", round(df["Sum of Dec"].mode().iloc[0], 2))
print("Minimum:", round(df["Sum of Dec"].min(), 2))
print("Maximum:", round(df["Sum of Dec"].max(), 2))
print("Range:", round(df["Sum of Dec"].max() - df["Sum of Dec"].min(), 2))
print("Variance:", round(df["Sum of Dec"].var(), 2))
print("Standard Deviation:", round(df["Sum of Dec"].std(), 2))
```

```python
print("Sum of May")

print("Count:", df["Sum of May"].count())
print("Mean:", round(df["Sum of May"].mean(), 2))
print("Median:", round(df["Sum of May"].median(), 2))
print("Mode:", round(df["Sum of May"].mode().iloc[0], 2))
print("Minimum:", round(df["Sum of May"].min(), 2))
print("Maximum:", round(df["Sum of May"].max(), 2))
print("Range:", round(df["Sum of May"].max() - df["Sum of May"].min(), 2))
print("Variance:", round(df["Sum of May"].var(), 2))
print("Standard Deviation:", round(df["Sum of May"].std(), 2))
```

## 8. Interactive histogram

A **histogram** shows the distribution of a numeric variable.

It groups values into intervals called **bins** and counts how many observations fall into each interval.


```python
print("Choose a variable:")
print("1 = Sum of Dec")
print("2 = Sum of May")

choice = input("Enter 1 or 2: ")

if choice == "1":
    variable = "Sum of Dec"
elif choice == "2":
    variable = "waiting"
else:
    print("Invalid choice. Using Sum of Dec.")
    variable = "Sum of Dec"

bins = input("How many bins would you like? Press Enter for 10: ")

if bins == "":
    bins = 10
else:
    bins = int(bins)

plt.figure(figsize=(9, 5))
plt.hist(df[variable], bins=bins, edgecolor="black")

plt.title(f"Distribution of {variable.title()}")
plt.xlabel(variable.title())
plt.ylabel("Frequency")

plt.show()
```

## 11. Create an observation number

This dataset does **not** contain a date or time variable.

For a line chart, we can create an observation number:

1, 2, 3, 4, ... through the final observation.

### Important

This means the line chart shows the **sequence of observations**, not change over calendar time.


```python
df["observation"] = range(1, len(df) + 1)

df.head()
```

## 12. Interactive line chart

Choose which numeric variable you would like to plot across observation order.


```python
print("Choose a variable for the line chart:")
print("1 = Sum of Dec")
print("2 = Sum of May")

choice = input("Enter 1 or 2: ")

if choice == "1":
    variable = "Sum of Dec"
elif choice == "2":
    variable = "Sum of May"
else:
    print("Invalid choice. Using Sum of Dec.")
    variable = "Sum of Dec"

plt.figure(figsize=(11, 5))

plt.plot(
    df["observation"],
    df[variable],
    marker="o",
    markersize=3
)

plt.title(f"{variable.title()} by Observation Number")
plt.xlabel("Observation Number")
plt.ylabel(variable.title())

plt.show()
```

## 13. Line chart of eruption duration


```python
plt.figure(figsize=(11, 5))

plt.plot(
    df["observation"],
    df["Sum of Dec"]
)

plt.title("Sum of Dec Duration by Observation Number")
plt.xlabel("Observation Number")
plt.ylabel("Sum of Dec Duration")

plt.show()
```

## 15. Final comparison table

Let's compare the two original numeric variables one more time.


```python
comparison = pd.DataFrame({
    "Mean": df[["Sum of Dec", "Sum of May"]].mean(),
    "Median": df[["Sum of Dec", "Sum of May"]].median(),
    "Minimum": df[["Sum of Dec", "Sum of May"]].min(),
    "Maximum": df[["Sum of Dec", "Sum of May"]].max(),
    "Range": (
        df[["Sum of Dec", "Sum of May"]].max()
        - df[["Sum of Dec", "Sum of May"]].min()
    ),
    "Standard Deviation": df[["Sum of Dec", "Sum of May"]].std()
})

comparison.round(2)
```
