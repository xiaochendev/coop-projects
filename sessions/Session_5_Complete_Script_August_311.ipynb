### 6G. Count complaint types within each borough

Now we move from one categorical variable to **two** categorical variables:

- Borough
- Problem type

We will group rows by both variables and count how many records fall into each combination.


````python
borough_problem_counts = (
    borough_df
    .groupby(["Borough", problem_col])
    .size()
    .reset_index(name="Count")
)

borough_problem_counts.head(10)
````

### What is `groupby()` doing?

Think of it as:

> **Split the rows into groups → calculate something for each group → return the results.**

Here the groups are combinations such as:

- Bronx + Illegal Parking
- Brooklyn + Illegal Parking
- Queens + Noise - Residential
- Manhattan + Encampment

`.size()` counts the rows in each group.

`.reset_index(name="Count")` turns the result back into a regular table and names the new count column `Count`.

> **Instructor note:** Students do not need to memorize this syntax yet. The goal is to understand the logic.


### 6H. Find the top five problems in each borough

We will rank the problem counts **within each borough**, then keep the top five.


````python
borough_problem_counts["Rank"] = (
    borough_problem_counts
    .groupby("Borough")["Count"]
    .rank(method="first", ascending=False)
)

top5_by_borough = (
    borough_problem_counts[borough_problem_counts["Rank"] <= 5]
    .sort_values(["Borough", "Rank"])
)

top5_by_borough
````

### 6I. Add percentages **within each borough**

Counts can be hard to compare when groups are different sizes.

For example:

- Brooklyn has many more total requests than Staten Island.
- Therefore, Brooklyn will often have larger raw counts even when the category is not a larger **share** of borough requests.

We can calculate a within-borough percentage:

> **Complaint count ÷ total requests in that borough × 100**


````python
borough_totals = (
    borough_df
    .groupby("Borough")
    .size()
    .reset_index(name="Borough Total")
)

borough_problem_pct = borough_problem_counts.merge(
    borough_totals,
    on="Borough"
)

borough_problem_pct["Percent within Borough"] = (
    borough_problem_pct["Count"]
    / borough_problem_pct["Borough Total"]
    * 100
).round(1)

borough_problem_pct.head()
````

### 6J. Display the top five with counts **and** percentages

Now we can answer a more useful question:

> Within each borough, what share of all 311 requests came from each leading complaint type?


````python
borough_problem_pct["Rank"] = (
    borough_problem_pct
    .groupby("Borough")["Count"]
    .rank(method="first", ascending=False)
)

top5_with_pct = (
    borough_problem_pct[borough_problem_pct["Rank"] <= 5]
    [["Borough", problem_col, "Count", "Percent within Borough"]]
    .sort_values(["Borough", "Count"], ascending=[True, False])
)

top5_with_pct
````

`pd.crosstab()` creates a table that counts combinations of categories.

We will look at the five most common problem types citywide and compare their counts across boroughs.


````python
top5_citywide = df[problem_col].value_counts().head(5).index

borough_problem_table = pd.crosstab(
    borough_df["Borough"],
    borough_df[problem_col]
)

borough_problem_table[top5_citywide]
````

### Why is this table useful?

A cross-tabulation makes side-by-side comparison easier.

Ask students to scan **across a row** and then **down a column**:

- Across a row: What complaints are most common within one borough?
- Down a column: Which borough has the highest count for one complaint type?

> **Connection to Excel:** A cross-tabulation is conceptually similar to a PivotTable that counts records by row and column categories.


### 6L. One simple chart: borough request counts

Tables are excellent for exact values. Charts can make broad differences easier to see.

We can create a simple bar chart directly from the borough counts.


````python
borough_counts.loc[nyc_boroughs].sort_values().plot(
    kind="barh",
    title="August 2026 NYC 311 Requests by Borough",
    xlabel="Number of Requests",
    ylabel="Borough",
    figsize=(8, 5)
)
````

### Read the chart before interpreting it

Ask:

1. What variable is on the vertical axis?
2. What variable is on the horizontal axis?
3. Which bar is longest?
4. Does the chart show a **count** or a **rate**?

> **Important:** This is a count chart. We should not call it a per-capita comparison.


### 6M. One more table: request status

A final quick example reinforces that the same method works on another categorical variable.


````python
status_summary = pd.DataFrame({
    "Count": df["Status"].value_counts(),
    "Percent": (df["Status"].value_counts(normalize=True) * 100).round(1)
})

status_summary
````

### Checkpoint

Most records are listed as **Closed**. You should see approximately:

- Closed: 272,806
- Open: 28,112
- In Progress: 26,081

### Ask the class

Can we conclude from this table alone that agencies handled requests quickly?

**No.** Status tells us the current category, but it does not tell us how long closure took. For that, we would need to work with `Created Date` and `Closed Date` as dates and calculate elapsed time.

This is a useful example of turning an answer into the **next analytical question**.
